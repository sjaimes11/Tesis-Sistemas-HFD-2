# Analisis SRE - hfl_v7-RN (#3 a #11)

Este notebook construye una capa de observabilidad SRE sobre los CSV historicos ya guardados en `hfl_v7-RN/Results`.

No modifica los archivos fuente del experimento. Solo genera tablas derivadas, trazas de ronda, catalogos SRE y figuras tipo dashboard en `analysis_outputs/sre_results_3_11`.


```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import Markdown, Image, display

NOTEBOOK_DIR = Path(r"C:/Users/VivoBook/Downloads/microproyecto2/microproyecto 2 porque wtf/src/Analisis de Modelos/RN")
RESULTS_ROOT = Path(r"C:/Users/VivoBook/Downloads/microproyecto2/microproyecto 2 porque wtf/src/hfl_v7-RN/Results")
OUTPUT_ROOT = NOTEBOOK_DIR / "analysis_outputs"
ATTEMPT_START = 3
ATTEMPT_END = 11
EXPECTED_ROUNDS_PER_ATTEMPT = 30

if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

import sre_hfl_rn_analysis as sre

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 220)
pd.set_option("display.max_rows", 100)
```

```python
bundle = sre.run_analysis(
    results_root=RESULTS_ROOT,
    output_root=OUTPUT_ROOT,
    attempt_start=ATTEMPT_START,
    attempt_end=ATTEMPT_END,
    expected_rounds_per_attempt=EXPECTED_ROUNDS_PER_ATTEMPT,
)

display(Markdown(f"## Output Dir\n`{bundle['output_dir']}`"))
display(pd.DataFrame([bundle["executive_summary"]]))
```

```python
display(Markdown("## Inventario de Intentos"))
display(bundle["tables"]["attempt_inventory"])

display(Markdown("## Metricas Exactas"))
display(bundle["tables"]["metric_catalog"])

display(Markdown("## Logs Exactos / Live Only"))
display(bundle["tables"]["log_catalog"])

display(Markdown("## Trazas Exactas / Live Only"))
display(bundle["tables"]["trace_catalog"])

display(Markdown("## Dashboard SRE"))
display(bundle["tables"]["dashboard_catalog"])
```

```python
display(Markdown("## SLI de Transporte"))
display(bundle["tables"]["transport_sli_summary"].head(20))

display(Markdown("## SLI de Entrenamiento Local"))
display(bundle["tables"]["local_training_sli_summary"])

display(Markdown("## SLI Global por Intento"))
display(bundle["tables"]["global_round_sli_summary"])

display(Markdown("## Alineacion entre Gateways"))
display(bundle["tables"]["gateway_alignment_sli_summary"].head(20))

display(Markdown("## Trazas Reconstruidas por Ronda"))
display(bundle["tables"]["round_trace_summary"].head(20))
```
