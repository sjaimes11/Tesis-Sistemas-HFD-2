```python
from pathlib import Path
import os
import sys

def find_ml_classic_dir(script_name: str) -> Path:
    for base in [Path.cwd(), *Path.cwd().parents]:
        if (base / script_name).exists() and (base / "data_loader.py").exists():
            return base
        candidate = base / "ml_classic"
        if (candidate / script_name).exists() and (candidate / "data_loader.py").exists():
            return candidate
    raise FileNotFoundError("No se encontró la carpeta ml_classic")

ML_CLASSIC_DIR = find_ml_classic_dir("decision_tree_model.py")
os.chdir(ML_CLASSIC_DIR)
if str(ML_CLASSIC_DIR) not in sys.path:
    sys.path.insert(0, str(ML_CLASSIC_DIR))

print(f"Usando ml_classic desde: {ML_CLASSIC_DIR}")
```
