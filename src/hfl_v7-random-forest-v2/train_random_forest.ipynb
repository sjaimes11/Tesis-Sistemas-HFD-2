# HFL v7 - Random Forest

Este notebook entrena un teacher clasico y lo destila a una MLP compatible con `hfl_v7-RN`.
El resultado se guarda en esta misma carpeta como `ids_3class.keras` y `model_weights.h`.


```python
from pathlib import Path
import os
import sys

def find_repo_root(start: Path) -> Path:
    explicit = os.environ.get('HFL_REPO_ROOT') or os.environ.get('ML_CLASSIC_REPO_ROOT')
    if explicit:
        candidate = Path(explicit).expanduser().resolve()
        if candidate.exists():
            return candidate

    markers = ('ml_classic', 'ml_classic_v2')
    search_roots = [start, *start.parents]
    for candidate in search_roots:
        if any((candidate / marker).exists() for marker in markers):
            return candidate

    for candidate in search_roots:
        for marker in markers:
            marker_dir = candidate / marker
            if (marker_dir / 'hfl_teacher_student_bundle.py').exists():
                return candidate

    raise RuntimeError('No se pudo localizar la raiz del proyecto.')

REPO_ROOT = find_repo_root(Path.cwd().resolve())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

if (REPO_ROOT / 'ml_classic').exists():
    from ml_classic.hfl_teacher_student_bundle import train_teacher_student_bundle
elif (REPO_ROOT / 'ml_classic_v2').exists():
    from ml_classic_v2.hfl_teacher_student_bundle import train_teacher_student_bundle
else:
    raise RuntimeError('No se encontro ni ml_classic ni ml_classic_v2 dentro de la raiz detectada.')

if (REPO_ROOT / 'Data').exists():
    DATA_DIR = str(REPO_ROOT / 'Data')
elif (REPO_ROOT / 'Data_Sets').exists():
    DATA_DIR = str(REPO_ROOT / 'Data_Sets')
else:
    DATA_DIR = '/home/sajaimesp/Tesis_Sistemas/Data_Sets'
BUNDLE_DIR = str(REPO_ROOT / 'hfl_v7-random-forest')
MODEL_KIND = 'random_forest'
RANDOM_STATE = 42
TEST_SIZE = 0.2
MAX_SAMPLES_PER_CLASS = None
SAVE_NATIVE_KERAS = False  # En Linux puedes ponerlo en True.
```

```python
summary = train_teacher_student_bundle(
    model_kind=MODEL_KIND,
    data_dir=DATA_DIR,
    bundle_dir=BUNDLE_DIR,
    random_state=RANDOM_STATE,
    test_size=TEST_SIZE,
    max_samples_per_class=MAX_SAMPLES_PER_CLASS,
    save_native_keras=SAVE_NATIVE_KERAS,
)
summary
```
