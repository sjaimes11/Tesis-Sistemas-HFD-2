# HFL v7 - K-Nearest Neighbors

Entrena un teacher clasico y lo destila a una MLP compatible con HFL v7.


```python
from pathlib import Path
import sys

ML_CLASSIC_DIR = Path('/home/sajaimesp/Tesis_Sistemas/Data_Sets/ml_classic').resolve()
DATA_DIR = ML_CLASSIC_DIR.parent

if str(ML_CLASSIC_DIR) not in sys.path:
    sys.path.insert(0, str(ML_CLASSIC_DIR))

from hfl_teacher_student_bundle import train_teacher_student_bundle

MODEL_KIND = 'knn'
BUNDLE_DIR = str(DATA_DIR / 'hfl_v7-knn')
RANDOM_STATE = 42
TEST_SIZE = 0.2
MAX_SAMPLES_PER_CLASS = 30000
SAVE_NATIVE_KERAS = True
```

```python
summary = train_teacher_student_bundle(
    model_kind=MODEL_KIND,
    data_dir=str(DATA_DIR),
    bundle_dir=BUNDLE_DIR,
    random_state=RANDOM_STATE,
    test_size=TEST_SIZE,
    max_samples_per_class=MAX_SAMPLES_PER_CLASS,
    save_native_keras=SAVE_NATIVE_KERAS,
)
summary
```
